# UNITTS quickstart

Hear an open-source text-to-speech model in about two minutes, on a free CPU runtime.
This installs UNITTS with the [Kokoro](https://github.com/hexgrad/kokoro) engine, exactly as
`scripts/setup.py` does on your own machine, then synthesizes a sentence.

Repository: https://github.com/aynursusuz/UNITTS

````python
# System package for Kokoro's phonemizer, the uv installer, and the repository.
!apt-get -qq install -y espeak-ng > /dev/null
!pip -q install uv
!git clone -q https://github.com/aynursusuz/UNITTS.git
%cd UNITTS
!python3 scripts/setup.py --engine kokoro --device cpu
````

## Synthesize

The engine runs in the environment that setup created (`.venv-kokoro`). The first run
downloads the Kokoro weights (about 330 MB).

````python
!.venv-kokoro/bin/unitts synthesize "One Python interface for every text-to-speech model." --engine kokoro --output kokoro.wav

from IPython.display import Audio
Audio("kokoro.wav")
````

## Another engine, same call

Supertonic also runs on CPU. Only the engine name changes.

````python
!python3 scripts/setup.py --engine supertonic --device cpu
!.venv-supertonic/bin/unitts synthesize "One Python interface for every text-to-speech model." --engine supertonic --output supertonic.wav
Audio("supertonic.wav")
````

## In Python

```python
from unitts.engines import get_engine

engine = get_engine("kokoro")
engine.synthesize_to_file("Hello world!", "out.wav")
```

The other engines (Chatterbox, Qwen3-TTS, Fish Audio, Echo-TTS, NeuTTS, MOSS-TTS, Higgs)
are listed in the [README](https://github.com/aynursusuz/UNITTS#engines); several of them need
an NVIDIA GPU runtime.
